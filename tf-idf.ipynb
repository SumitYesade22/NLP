```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
    "I really love this smartphone! The camera quality is excellent.",
    "This smartphone has a very good camera, but the battery is not great.",
    "I do not like this phone because it is too expensive.",
    "The battery life is amazing and the camera is also very good.",
    "This phone is not worth buying. It has poor performance and bad battery life.",
    "I love the performance of this smartphone, but its camera is average.",
    "The phone costs Rs. 25,000 and offers 8GB RAM with 128GB storage.",
    "Is this smartphone better than the previous model? I am not sure.",
    "The camera camera camera is excellent, but the phone is expensive.",
    "I purchased this phone in 2025. It works well for gaming and photography."
]
```

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(documents)
```

```python
feature_names = tfidf.get_feature_names_out()

print("Total vocabulary:", len(feature_names))
print(feature_names)
```

```python
idf_df = pd.DataFrame({
    "Word": feature_names,
    "IDF": tfidf.idf_
})

idf_df = idf_df.sort_values("IDF", ascending=True)

idf_df
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names,
    index=[f"Document_{i}" for i in range(1, len(documents) + 1)]
)

tfidf_df
```
