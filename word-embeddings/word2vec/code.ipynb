```python
from gensim.models import Word2Vec
sentences = [
    "I love machine learning",
    "I love deep learning",
    "machine learning is interesting",
    "deep learning is powerful",
    "I enjoy learning Python",
    "Python is useful for machine learning"
]
tokenized_sentences = [
    sentence.lower().split()
    for sentence in sentences
]

tokenized_sentences
```

```python
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=0,
    epochs=100
)
```

```python
print(model)
```

```python
print(model.wv.index_to_key)
```

```python
model.wv["machine"]
```

```python
sg_model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,
    epochs=100
)
```

```python
sg_model.wv["machine"]
```

```python
from gensim.models import FastText
fasttext_model = FastText(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,
    epochs=100,
    min_n=3,
    max_n=6
)
fasttext_model.wv["machine"]
```
