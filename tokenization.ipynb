```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Natural Language Processing is a field of Artificial Intelligence.
It enables computers to understand and process human language.
I am learning NLP using Python and spaCy.
"""
```

```python
doc = nlp(text)

print(doc)
print(type(doc))
```

```python
tokens = [token.text for token in doc]

print(tokens)
```

```python
for token in doc:
    print(
        repr(token.text),
        "is_space:", token.is_space,
        "whitespace_after:", repr(token.whitespace_)
    )
```

```python
sentences = [sent.text for sent in doc.sents]

print(sentences)
```

```python
sentences = [
    sent.text.strip()
    for sent in doc.sents
]

print(sentences)
```

```python
characters = [
    char
    for token in doc
    for char in token.text
]

print(characters)
```

```python
tokens = [
    token.text
    for token in doc
    if not token.is_space
]

print(tokens)
```

```python
tokens = [
    token.text
    for token in doc
    if not token.is_punct and not token.is_space
]

print(tokens)
```

```python
text = "I purchased 2 phones for 25000 rupees."

doc = nlp(text)

tokens = [
    token.text
    for token in doc
    if not token.is_space
    and not token.is_punct
    and not token.is_stop
    and not token.like_num
]

print(tokens)
```
