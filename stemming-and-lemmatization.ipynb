```python
import nltk
import spacy

from nltk.stem import PorterStemmer, SnowballStemmer
```

```python
text = """
The children are playing in the gardens.
The boys played football and were running.
The studies were interesting and the students studied carefully.
The mice were eating better food.
The cars are connected to different devices.
The companies are developing new technologies.
"""
```

```python
porter = PorterStemmer()

snowball = SnowballStemmer("english")

nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(text)
```

```python
for token in doc:
    if not token.is_space and not token.is_punct:
        print(token.text, token.pos_, token.lemma_)
```

```python
porter_results = []

for token in doc:
    if token.is_alpha:
        stem = porter.stem(token.text.lower())
        porter_results.append((token.text, stem))

print(porter_results)
```

```python
snowball_results = []

for token in doc:
    if token.is_alpha:
        stem = snowball.stem(token.text.lower())
        snowball_results.append((token.text, stem))

print(snowball_results)
```

```python
lemma_results = []

for token in doc:
    if token.is_alpha:
        lemma_results.append((token.text, token.lemma_))

print(lemma_results)
```
