```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
text = """
I am learning Natural Language Processing.
This is an interesting field and I want to understand it properly.
I do not want to miss important concepts.
The model is learning from the data.
"""
```

```python
doc = nlp(text)
```

```python
for token in doc:
    if not token.is_space:
        print(token.text, "->", token.is_stop)
```

```python
stopwords = [
    token.text
    for token in doc
    if token.is_stop
]

print(stopwords)
```

```python
filtered_tokens = [
    token.text
    for token in doc
    if not token.is_stop and not token.is_space
]

print(filtered_tokens)
```
